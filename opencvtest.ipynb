```python
cap= cv2.VideoCapture(0)#获取图像
cap.set(3,640)#3：长
cap.set(4,480)#4：宽
while True:
    success,img=cap.read()

    if not success:
        break
    cv2.imshow("Video",img)
    if cv2.waitKey(8) & 0xFF==ord('q'):
     break
cap.release()  # 释放摄像头，不释放下次开摄像头会 busy
cv2.destroyAllWindows()  # 关掉所有 imshow 弹出来的窗口
cv2.waitKey(1)  # 让窗口系统处理一次事件，防止窗口卡住不关
```

```python
import cv2
img=cv2.imread("Resources/lean.png")
imgGray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)  # BGR 三通道 -> 单通道灰度
imgBlur=cv2.GaussianBlur(img,(15,15),0)#转化为灰度图像
imgCanny=cv2.Canny(img,100,100)#把图像边缘化（把色彩扣去）
cv2.imshow("Gray Image",imgGray)  # 弹窗显示灰度图
cv2.imshow("Blur",imgBlur)
cv2.imshow("Canny",imgCanny)
cv2.waitKey(0)  # 阻塞等按键，按任意键关窗继续；0 表示无限等  就
```
